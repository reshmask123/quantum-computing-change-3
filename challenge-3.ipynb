```python
# Qiskit Fall Fest 2026
# Challenge 3: Can You Change the Correlation?

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create simulator
simulator = AerSimulator()

# ==========================================
# PART A: ORIGINAL BELL STATE CIRCUIT
# ==========================================

qc_original = QuantumCircuit(2, 2)

# Create Bell state
qc_original.h(0)
qc_original.cx(0, 1)

# Measure both qubits
qc_original.measure([0, 1], [0, 1])

print("ORIGINAL BELL STATE CIRCUIT")
display(qc_original.draw(output="mpl"))

# Run for 100 shots
job_original = simulator.run(qc_original, shots=100)
result_original = job_original.result()
counts_original = result_original.get_counts()

print("Original Circuit Results:")
print(counts_original)

# ==========================================
# PART B: MODIFIED CIRCUIT
# ==========================================

qc_modified = QuantumCircuit(2, 2)

# Create Bell state
qc_modified.h(0)
qc_modified.cx(0, 1)

# Apply X gate to qubit 0
qc_modified.x(0)

# Measure both qubits
qc_modified.measure([0, 1], [0, 1])

print("\nMODIFIED CIRCUIT")
display(qc_modified.draw(output="mpl"))

# Run for 100 shots
job_modified = simulator.run(qc_modified, shots=100)
result_modified = job_modified.result()
counts_modified = result_modified.get_counts()

print("Modified Circuit Results:")
print(counts_modified)

# ==========================================
# PART C: HISTOGRAM COMPARISON
# ==========================================

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

plot_histogram(
    counts_original,
    ax=axes[0],
    title="Original Bell State"
)

plot_histogram(
    counts_modified,
    ax=axes[1],
    title="Modified Bell State"
)

plt.tight_layout()
plt.show()

# ==========================================
# PART D: EXPLANATION
# ==========================================

print("\nEXPLANATION")
print("Original circuit: Outcomes 00 and 11 dominate.")
print("Modified circuit: Outcomes 01 and 10 dominate.")
print("The X gate flips qubit 0 and changes the correlation pattern.")
```
